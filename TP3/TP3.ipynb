## TP3.1 a) Démonstration de l'orthogonalité des matrices $Q_i$

## Étape 1 : Montrer que $H_{m,i}$ est orthogonale

Pour simplifier la notation, posons $H = H_{m,i}$, $I = I_{m-i}$, et $v = v_{m,i}$. La définition devient :
$$
H = I - 2 \frac{v v^T}{v^T v}
$$

Nous devons montrer que $H^T H = I$.

---

### Étape 1.1 : Calcul de la transposée $H^T$

$$
H^T = \left( I - 2 \frac{v v^T}{v^T v} \right)^T
$$

En utilisant les propriétés de la transposée :  
1. $(A - B)^T = A^T - B^T$,  
2. $(cA)^T = c A^T$,  
3. $(AB)^T = B^T A^T$, on obtient :

$$
H^T = I^T - \frac{2}{v^T v} (v v^T)^T
$$

- Comme $I$ est la matrice identité, $I^T = I$.  
- Le terme $v^T v$ est un scalaire.  
- La transposée du produit extérieur est $(v v^T)^T = (v^T)^T v^T = v v^T$.  

Ainsi :
$$
H^T = I - 2 \frac{v v^T}{v^T v}
$$

On observe que $H^T = H$.  
**La matrice de Householder $H_{m,i}$ est symétrique.**

---

### Étape 1.2 : Calcul du produit $H^T H$

Nous devons calculer $H H$, où $H$ est défini comme :
$$
H = I - 2 \frac{v v^T}{v^T v}.
$$

En développant le produit $H H$ :
$$
H H = \left( I - 2 \frac{v v^T}{v^T v} \right) \left( I - 2 \frac{v v^T}{v^T v} \right).
$$

#### Développement avec la distributivité :
$$
H H = I - 2 \frac{v v^T}{v^T v} - 2 \frac{v v^T}{v^T v} + 4 \frac{(v v^T)(v v^T)}{(v^T v)^2}.
$$

#### Analyse des termes :
1. **Premier terme :**  
   $I$, la matrice identité, reste inchangé.

2. **Deuxième et troisième termes :**  
   Ces deux termes sont identiques et se simplifient ensemble :
   $$
   - 2 \frac{v v^T}{v^T v} - 2 \frac{v v^T}{v^T v} = - 4 \frac{v v^T}{v^T v}.
   $$

3. **Dernier terme (quadratique) :**  
   Le dernier terme est :
   $$
   4 \frac{(v v^T)(v v^T)}{(v^T v)^2}.
   $$
   En utilisant l'associativité de la multiplication matricielle, on observe que :
   $$
   (v v^T)(v v^T) = v (v^T v) v^T.
   $$
   Donc :
   $$
   4 \frac{(v v^T)(v v^T)}{(v^T v)^2} = 4 \frac{v (v^T v) v^T}{(v^T v)^2}.
   $$
   En simplifiant, cela devient :
   $$
   4 \frac{v v^T}{v^T v}.
   $$

#### Substitution des termes :
En remplaçant ces simplifications dans l'expression de $H H$, on obtient :
$$
H H = I - 4 \frac{v v^T}{v^T v} + 4 \frac{v v^T}{v^T v}.
$$

Les deux derniers termes s'annulent :
$$
H H = I.
$$

---

### Conclusion :
Nous avons montré que :
$$
H^T H = H H = I.
$$
Ainsi, **la matrice de Householder $H_{m,i}$ est orthogonale.**

---

## Étape 2 : Montrer que $Q_i$ est orthogonale

Rappelons la structure par blocs de $Q_i$ (taille $m \times m$) :
$$
Q_i =
\begin{bmatrix}
I_i & 0 \\
0 & H_{m,i}
\end{bmatrix}
$$

Nous devons montrer que $Q_i^T Q_i = I_m$, où $I_m$ est la matrice identité de taille $m \times m$.

---

### Étape 2.1 : Calcul de la transposée $Q_i^T$

En utilisant la règle de transposition des matrices par blocs :
$$
Q_i^T =
\begin{bmatrix}
I_i^T & 0^T \\
0^T & H_{m,i}^T
\end{bmatrix}
$$

- $I_i^T = I_i$, car $I_i$ est une matrice identité.  
- $H_{m,i}^T = H_{m,i}$, car $H_{m,i}$ est symétrique.  

Ainsi :
$$
Q_i^T =
\begin{bmatrix}
I_i & 0 \\
0 & H_{m,i}
\end{bmatrix}
= Q_i
$$

**La matrice $Q_i$ est également symétrique.**

---

### Étape 2.2 : Calcul du produit $Q_i^T Q_i$

Puisque $Q_i^T = Q_i$, nous calculons $Q_i Q_i$ :
$$
Q_i Q_i =
\begin{bmatrix}
I_i & 0 \\
0 & H_{m,i}
\end{bmatrix}
\begin{bmatrix}
I_i & 0 \\
0 & H_{m,i}
\end{bmatrix}
$$

En utilisant la multiplication des matrices par blocs :
$$
Q_i Q_i =
\begin{bmatrix}
I_i^2 & 0 \\
0 & H_{m,i}^2
\end{bmatrix}
$$

- $I_i^2 = I_i$, car $I_i$ est une matrice identité.  
- $H_{m,i}^2 = I_{m-i}$ (démontré à l’Étape 1).  

Ainsi :
$$
Q_i Q_i =
\begin{bmatrix}
I_i & 0 \\
0 & I_{m-i}
\end{bmatrix}
= I_m
$$

---

### Conclusion :
Nous avons montré que $Q_i^T = Q_i$ et que $Q_i^T Q_i = Q_i Q_i = I_m$.  
Par définition, **les matrices de réflexion $Q_i$ sont orthogonales.**

## TP3.1 b) Démonstration de l'équation (3.5) et de l'orthogonalité de $Q$

### Objectifs :
1. Démontrer l'équation (3.5) :  
   $Q = Q_0^T Q_1^T \dots Q_{n-1}^T$
2. Démontrer que cette matrice $Q$ est orthogonale.

---

## 1. Démonstration de l'équation (3.5)

On part de l'équation fondamentale du processus de Householder (Éq. 3.1) :
$$
Q_{n-1} Q_{n-2} \dots Q_1 Q_0 A = R
$$

Nous voulons réarranger cette équation pour obtenir la forme standard de la décomposition QR : $A = QR$.  
Pour ce faire, nous devons isoler $A$ en multipliant successivement par les inverses des matrices $Q_i$ à gauche.

D'après la partie (a), nous avons démontré que chaque matrice $Q_i$ est orthogonale.  
Une propriété clé des matrices orthogonales est que leur inverse est égal à leur transposée :
$$
Q_i^{-1} = Q_i^T
$$

Multiplions l'équation (3.1) par $Q_{n-1}^{-1}$ à gauche :
$$
Q_{n-1}^{-1} \left(Q_{n-1} Q_{n-2} \dots Q_1 Q_0 A \right) = Q_{n-1}^{-1} R
$$
$$
\left(Q_{n-1}^{-1} Q_{n-1}\right) Q_{n-2} \dots Q_1 Q_0 A = Q_{n-1}^{-1} R
$$
$$
I Q_{n-2} \dots Q_1 Q_0 A = Q_{n-1}^{-1} R
$$
$$
Q_{n-2} \dots Q_1 Q_0 A = Q_{n-1}^{-1} R
$$

Multiplions maintenant par $Q_{n-2}^{-1}$ à gauche :
$$
Q_{n-2}^{-1} \left(Q_{n-2} \dots Q_1 Q_0 A\right) = Q_{n-2}^{-1} Q_{n-1}^{-1} R
$$
$$
Q_{n-3} \dots Q_1 Q_0 A = Q_{n-2}^{-1} Q_{n-1}^{-1} R
$$

En continuant ce processus $n$ fois (de $i = n-1$ jusqu'à $i = 0$), on obtient :
$$
A = Q_0^{-1} Q_1^{-1} \dots Q_{n-2}^{-1} Q_{n-1}^{-1} R
$$

Maintenant, substituons $Q_i^{-1} = Q_i^T$ pour chaque $i$ :
$$
A = \left(Q_0^T Q_1^T \dots Q_{n-2}^T Q_{n-1}^T\right) R
$$

En comparant cette équation avec la forme standard $A = QR$, nous identifions la matrice $Q$ comme :
$$
Q = Q_0^T Q_1^T \dots Q_{n-1}^T
$$

Ceci démontre l'équation (3.5).

---

## 2. Démonstration de l'orthogonalité de $Q$

Pour montrer que la matrice $Q = Q_0^T Q_1^T \dots Q_{n-1}^T$ est orthogonale, nous devons prouver que $Q^T Q = I$, où $I$ est la matrice identité.

### Étape 2.1 : Calcul de $Q^T$
En utilisant la propriété de la transposée d'un produit de matrices $(ABC)^T = C^T B^T A^T$, nous avons :
$$
Q^T = \left(Q_0^T Q_1^T \dots Q_{n-1}^T\right)^T
$$
$$
Q^T = \left(Q_{n-1}^T\right)^T \dots \left(Q_1^T\right)^T \left(Q_0^T\right)^T
$$

Sachant que $\left(M^T\right)^T = M$ pour toute matrice $M$, on obtient :
$$
Q^T = Q_{n-1} \dots Q_1 Q_0
$$

---

### Étape 2.2 : Calcul de $Q^T Q$
Calculons maintenant le produit $Q^T Q$ :
$$
Q^T Q = \left(Q_{n-1} \dots Q_1 Q_0\right) \left(Q_0^T Q_1^T \dots Q_{n-1}^T\right)
$$

Utilisons l'associativité de la multiplication matricielle pour regrouper les termes au centre :
$$
Q^T Q = Q_{n-1} \dots Q_1 \left(Q_0 Q_0^T\right) Q_1^T \dots Q_{n-1}^T
$$

Comme chaque $Q_i$ est orthogonale (prouvé en (a)), nous savons que $Q_i Q_i^T = I$ pour tout $i$. Donc, $Q_0 Q_0^T = I$ :
$$
Q^T Q = Q_{n-1} \dots Q_1 I Q_1^T \dots Q_{n-1}^T
$$
$$
Q^T Q = Q_{n-1} \dots \left(Q_1 Q_1^T\right) \dots Q_{n-1}^T
$$

De même, $Q_1 Q_1^T = I$ :
$$
Q^T Q = Q_{n-1} \dots Q_2 I Q_2^T \dots Q_{n-1}^T
$$
$$
Q^T Q = Q_{n-1} \dots \left(Q_2 Q_2^T\right) \dots Q_{n-1}^T
$$

En continuant ce processus d'élimination par paires $Q_i Q_i^T = I$ du centre vers l'extérieur, nous nous retrouvons finalement avec :
$$
Q^T Q = Q_{n-1} Q_{n-1}^T
$$

Et puisque $Q_{n-1}$ est orthogonale, $Q_{n-1} Q_{n-1}^T = I$.  
Donc :
$$
Q^T Q = I
$$

---

### Conclusion :
Nous avons montré que $Q^T Q = I$.  
Par définition, la matrice $Q$ définie par l'équation (3.5) est orthogonale.
